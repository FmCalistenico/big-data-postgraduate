**Reglas de estilo:** título con año, unidad en la leyenda, fuente al pie, municipios sin dato en gris con trama, cortes por cuantiles declarados en el pie. Se dibuja en EPSG:9377.

```python
import sys, pathlib
ROOT = pathlib.Path.cwd().parent if pathlib.Path.cwd().name == "notebooks" else pathlib.Path.cwd()
sys.path.insert(0, str(ROOT))
import polars as pl
from src.config import *
asegurar_carpetas()
```

```python
import geopandas as gpd
from src import mapas
g = gpd.read_parquet(GOLD / "maestro_boyaca.geoparquet")
F_POB, F_MDM, F_EVA = "DANE, proyecciones de población", "DNP, MDM", "UPRA, EVA"
F_IGAC = "IGAC, capacidad de uso 1:100.000 (2004)"
y1, y2 = ANIO_MDM, ANIO_EVA
```

```python
mapas.coropleta(g, f"prop_18_44_pct_{y1}", f"Población de 18 a 44 años, {y1}", "% del total", F_POB, "m1_poblacion_joven")
mapas.coropleta(g, f"puntaje_mdm_{y1}", f"Desempeño municipal (MDM), {y1}", "Puntaje 0-100", F_MDM, "m2_desempeno", cmap="PuRd")
mapas.coropleta(g, f"rendimiento_papa_t_ha_{y2}", f"Rendimiento de papa, {y2}", "t/ha", F_EVA, "m3_rendimiento_papa", cmap="YlGn")
mapas.coropleta(g, f"produccion_1k_18_64_{y2}", f"Producción agrícola por 1.000 hab. de 18-64 años, {y2}",
                "t por 1.000 hab.", f"{F_EVA}; {F_POB}", "m4_produccion_por_1000", cmap="YlOrBr")
mapas.categorico(g, "clase_predominante", "Clase de capacidad de uso predominante", F_IGAC, "m5_clase_predominante",
                 orden=["III", "IV", "V", "VI", "VII", "VIII"])
```

```python
mapas.bivariado(g, f"prop_18_44_pct_{y1}", f"puntaje_mdm_{y1}", f"Población joven y desempeño municipal, {y1}",
                "Joven", "MDM", f"{F_POB}; {F_MDM}", "m6_bivariado_joven_mdm")
mapas.lado_a_lado(g, [(f"prop_18_44_pct_{y1}", "Población de 18-44 años", "% del total"),
                      (f"puntaje_mdm_{y1}", "Desempeño municipal", "Puntaje 0-100")],
                  f"Población joven y desempeño municipal, {y1}", f"{F_POB}; {F_MDM}", "m7_lado_a_lado")
```
