## 1. Inspección inicial · 10 minutos

✅ **Comprobación:** El corte del manifiesto contiene 92.738 filas y 32 columnas. La tabla original debe conservar esa cantidad de filas durante el ejercicio.

## 2. Diagnóstico de cinco campos · 10 minutos

Construir una tabla llamada `perfil` para `Secuencial`, `Fecha de Análisis`, `Departamento`, `Municipio` y `pH agua:suelo`. Para cada campo se registran los nulos de pandas, los textos vacíos después de retirar espacios y los marcadores `ND`. Las tres situaciones se cuentan por separado.

📘 **Apoyo:** `isna`, `.str.strip`, `.str.upper`, `.eq` y `sum`.

✅ **Comprobación:** `perfil` debe tener una fila por campo. Con `keep_default_na=False`, un texto vacío o `ND` no se convierte automáticamente en un nulo.

````python
# Desarrollo de la actividad 2
campos = ["Secuencial", "Fecha de Análisis", "Departamento", "Municipio", "pH agua:suelo"]

filas = []
for campo in campos:
    columna = suelo[campo]
    limpio = columna.str.strip()
    filas.append({
        "campo": campo,
        "nulos_pandas": int(columna.isna().sum()),
        "vacios_texto": int(limpio.eq("").sum()),
        "marcador_ND": int(limpio.str.upper().eq("ND").sum()),
        "distintos": int(columna.nunique(dropna=False)),
    })

perfil = pd.DataFrame(filas).set_index("campo")
assert len(perfil) == len(campos)
display(perfil)
````

**Interpretación del resultado:**

`perfil` tiene una fila por cada uno de los cinco campos y separa tres situaciones distintas: nulos de pandas, textos vacíos y marcadores `ND`.

- **Nulos de pandas:** 0 en los cinco campos. Como se leyó con `keep_default_na=False`, pandas no crea `NaN` al leer: las ausencias quedan escondidas como texto. Si solo se usara `isna()`, el perfil diría que el archivo está completo.
- **Textos vacíos:** solo `Municipio` tiene **2 registros vacíos** (`""`). Esos 2 análisis no tienen municipio de procedencia y deberían revisarse antes de cualquier agregación municipal. Por eso `Municipio` muestra 991 valores distintos: 990 municipios más el texto vacío.
- **Marcadores `ND`:** 0 en los cinco campos.

Los demás campos están completos. `Secuencial` tiene 92.738 valores distintos, uno por fila, así que identifica de forma única cada análisis. `Fecha de Análisis` tiene 2.388 fechas distintas y `Departamento` 33 valores, sin ausencias. `pH agua:suelo` no tiene vacíos ni `ND` (737 valores distintos).

## 3. Conversión de pH y fecha · 15 minutos

Crear `ph` desde `pH agua:suelo` y `fecha_dt` desde `Fecha de Análisis`. Utilizar conversión numérica con `errors="coerce"` y fecha con formato `%d/%m/%Y`. Conservar los originales. Distinguir entradas vacías de textos no vacíos que no pudieron convertirse.

📘 **Apoyo:** `pd.to_numeric`, `.str.replace`, `pd.to_datetime`, `isna` y máscaras booleanas.

✅ **Comprobación:** Se muestran los conteos de conversiones fallidas y hasta cinco ejemplos con su valor original. No se reemplazan medidas ausentes por cero.

````python
# Desarrollo de la actividad 3
ph_texto = suelo["pH agua:suelo"].str.strip()
fecha_texto = suelo["Fecha de Análisis"].str.strip()

# Columnas nuevas; las originales no se modifican.
suelo["ph"] = pd.to_numeric(ph_texto.str.replace(",", ".", regex=False), errors="coerce")
suelo["fecha_dt"] = pd.to_datetime(fecha_texto, format="%d/%m/%Y", errors="coerce")

ph_vacio = ph_texto.eq("")
ph_no_numerico = ~ph_vacio & suelo["ph"].isna()
fecha_vacia = fecha_texto.eq("")
fecha_no_valida = ~fecha_vacia & suelo["fecha_dt"].isna()

fallos_conversion = pd.DataFrame({
    "entrada_vacia": [int(ph_vacio.sum()), int(fecha_vacia.sum())],
    "texto_no_convertible": [int(ph_no_numerico.sum()), int(fecha_no_valida.sum())],
}, index=["ph", "fecha_dt"])
fallos_conversion["total_ausente"] = [int(suelo["ph"].isna().sum()), int(suelo["fecha_dt"].isna().sum())]
display(fallos_conversion)

print("Ejemplos de pH no convertible (valor original):")
display(suelo.loc[ph_no_numerico, ["Secuencial", "pH agua:suelo"]].head(5))
print("Ejemplos de fecha no interpretable (valor original):")
display(suelo.loc[fecha_no_valida, ["Secuencial", "Fecha de Análisis"]].head(5))

print("Rango de pH convertido:", suelo["ph"].min(), "a", suelo["ph"].max())
print("Rango de fechas:", suelo["fecha_dt"].min(), "a", suelo["fecha_dt"].max())
assert len(suelo) == total_original
````

**Interpretación del resultado:**

Se crearon `ph` (numérico, cambiando la coma decimal por punto) y `fecha_dt` (fecha con formato explícito `%d/%m/%Y`), sin modificar las columnas originales `pH agua:suelo` y `Fecha de Análisis`.

- **pH:** 0 entradas vacías y **11 textos no convertibles**. No faltan datos: el valor existe pero no es un número. Los ejemplos muestran dos tipos de caso:
  - **Valores censurados**, como `>10.00` y `<3.80`: el laboratorio informa que el pH está por encima o por debajo del límite que mide el método, pero no da el valor exacto.
  - **Errores de digitación**, como `4..01` (doble punto), que probablemente correspondía a 4,01.
- **Fecha:** 0 vacías y 0 no interpretables. Todas las fechas siguen el formato día/mes/año.



## 4. Registros para revisión · 15 minutos

Crear la columna booleana `en_revision`. Se marca una fila si su pH es ausente, está fuera de `[0, 14]` o su fecha no se pudo interpretar. Contar por separado cada motivo y el total de filas marcadas. Conservar todas las observaciones.

📘 **Apoyo:** `between`, `isna`, `|`, `~`, `.loc` y `value_counts`.

✅ **Comprobación:** Las filas marcadas y no marcadas deben sumar el total original. Una misma fila puede tener varios motivos; por eso la suma de los motivos no necesariamente coincide con el total marcado.

````python
# Desarrollo de la actividad 4
motivo_ph_ausente = suelo["ph"].isna()
motivo_ph_fuera = suelo["ph"].notna() & ~suelo["ph"].between(0, 14)
motivo_fecha = suelo["fecha_dt"].isna()

suelo["en_revision"] = motivo_ph_ausente | motivo_ph_fuera | motivo_fecha

motivos = pd.Series({
    "pH ausente o no numérico": int(motivo_ph_ausente.sum()),
    "pH fuera de [0, 14]": int(motivo_ph_fuera.sum()),
    "fecha no interpretable": int(motivo_fecha.sum()),
}, name="filas")
display(motivos.to_frame())

conteo_revision = suelo["en_revision"].value_counts().rename("filas")
display(conteo_revision.to_frame())

marcadas = int(suelo["en_revision"].sum())
no_marcadas = int((~suelo["en_revision"]).sum())
print(f"Marcadas: {marcadas:,} | No marcadas: {no_marcadas:,} | Total: {marcadas + no_marcadas:,}")
print("Filas con más de un motivo:",
      int((motivo_ph_ausente.astype(int) + motivo_ph_fuera.astype(int) + motivo_fecha.astype(int)).gt(1).sum()))

assert marcadas + no_marcadas == total_original
assert len(suelo) == total_original
display(suelo.loc[suelo["en_revision"], ["Secuencial", "pH agua:suelo", "ph", "Fecha de Análisis", "fecha_dt"]].head(10))
````

**Interpretación del resultado:**

`en_revision` marca **11 filas** y deja **92.727 sin marca**. Su suma es 92.738, igual al total original, y el `assert` confirma que **no se eliminó ninguna observación**: las filas quedan señaladas para revisión, no borradas.

Por motivo: **11 por pH ausente o no numérico**, 0 por pH fuera de `[0, 14]` y 0 por fecha no interpretable. Hay **0 filas con más de un motivo**, así que en este caso la suma de motivos coincide con el total marcado; podría no coincidir si una misma fila tuviera, por ejemplo, pH y fecha inválidos a la vez.

La tabla de filas marcadas muestra que todas tienen fecha válida (entre 2014 y 2022) y que el problema está solo en el pH: valores censurados (`>10.00`, `<3.80`) y un error de digitación (`4..01`).

## 5. Evidencias y conclusiones · 10 minutos

Construir `controles` con las columnas `control` y `cantidad`: total de entrada, pH sin número, pH fuera del intervalo, fecha no interpretable y total de filas en revisión. Exportar a `OUT / "controles_agrosavia.csv"` sin índice. Redactar dos hallazgos apoyados en los conteos.

📘 **Apoyo:** `pd.DataFrame`, `to_csv(index=False)` y las tablas de las actividades anteriores.

✅ **Comprobación:** El archivo se genera dentro de la ruta `OUT`. Cada hallazgo menciona un conteo y explica su significado o limitación.

````python
# Desarrollo de la actividad 5
controles = pd.DataFrame({
    "control": [
        "total_entrada",
        "ph_sin_numero",
        "ph_fuera_intervalo",
        "fecha_no_interpretable",
        "total_en_revision",
    ],
    "cantidad": [
        total_original,
        int(motivo_ph_ausente.sum()),
        int(motivo_ph_fuera.sum()),
        int(motivo_fecha.sum()),
        int(suelo["en_revision"].sum()),
    ],
})
display(controles)

ruta = OUT / "controles_agrosavia.csv"
controles.to_csv(ruta, index=False, encoding="utf-8-sig")
assert ruta.is_file()
print("Archivo generado:", ruta)

pct = controles.loc[4, "cantidad"] / total_original * 100
print(f"Filas en revisión: {controles.loc[4, 'cantidad']:,} de {total_original:,} ({pct:.3f} %)")
````

**Interpretación del resultado:**

`controles` resume la conciliación del ejercicio (92.738 de entrada; 11 sin pH numérico; 0 fuera de intervalo; 0 fechas no interpretables; 11 en revisión) y se exportó a `kit/salidas/soluciones/PD01/controles_agrosavia.csv` sin índice.

**Hallazgo 1.** De 92.738 análisis, solo **11 (≈0,012 %)** no tienen un pH numérico utilizable, y ninguno está fuera del intervalo físico `[0, 14]`. La calidad del campo pH es muy alta: excluir esas filas casi no cambia el tamaño de la muestra. Sin embargo, la mayoría son **valores censurados** (`>10.00`, `<3.80`) que indican suelos extremadamente alcalinos o ácidos. Si se descartan sin más, se pierden justo los casos extremos. Que un valor pase el control `[0, 14]` no prueba que sea correcto agronómicamente; solo que es físicamente posible.

**Hallazgo 2.** Hay **0 fechas no interpretables**, y en los campos de identificación solo aparecen **2 municipios vacíos**; los problemas se concentran en pocos registros. Leer con `dtype=str` y `keep_default_na=False` fue necesario para verlos: con la lectura por defecto, los municipios vacíos se habrían convertido en `NaN` sin distinguir su origen, y los textos como `>10.00` se habrían mezclado con el resto de ausencias. Además, la cantidad de análisis por municipio refleja **dónde se tomaron muestras**, no la distribución real de los suelos del país.
